**Assumptions for Linear Regression**

<br>

| Assumption | Description | How to Check | Consequence if Violated |
|---|---|---|---|
| **Linearity** | Relationship between X and Y is linear | Scatterplot, Residuals vs Fitted plot | Biased/incorrect predictions |
| **Independence of Errors** | Residuals are not correlated with each other | Durbin-Watson test, know data collection process | Unreliable p-values (false significance) |
| **Homoscedasticity** | Constant variance of residuals across all X | Residuals vs Fitted plot (look for funnel shape) | Incorrect standard errors, unreliable p-values |
| **Normality of Residuals** | Residuals are approximately normally distributed | Q-Q plot, Histogram, Shapiro-Wilk test | Unreliable confidence intervals (mainly small samples) |
| **No Multicollinearity** | Independent variables aren't highly correlated with each other | VIF (Variance Inflation Factor), Correlation matrix | Unstable, hard-to-interpret coefficients |

<br><br>
**Additional Assumption Point**
<br>

| Assumption | Description | How to Check | Consequence if Violated |
|---|---|---|---|
| **No Significant Outliers** | No extreme points disproportionately influencing the fit | Cook's Distance, Leverage plots | Model skewed by a few points |
| **No Measurement Error in X** | Predictors are measured accurately | Domain knowledge, data collection review | Biased coefficient estimates |
