## Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
```

```python
s_scaler = StandardScaler()
```

```python
df['loan_amount_scaled'] = s_scaler.fit_transform(df[['loan_amount']])
df['loan_amount_scaled'].head()
```

```python
df[['loan_amount','loan_amount_scaled']].head()
```

```python
df['Upfront_charges_scaled'] = s_scaler.fit_transform(df[['Upfront_charges']])
df['Upfront_charges_scaled'].head()
```

```python
df[['Upfront_charges','Upfront_charges_scaled']]
```

## Min Max Scaler

```python
from sklearn.preprocessing import MinMaxScaler
m_scaler = MinMaxScaler()
df['loan_amount_Mscale'] = m_scaler.fit_transform(df[['loan_amount']])
```

```python
df[['loan_amount','loan_amount_scaled','loan_amount_Mscale']].head()
```

## Robust Scaler

```python
from sklearn.preprocessing import RobustScaler
r_scaler = RobustScaler()
df['loan_amount_Rscale'] = r_scaler.fit_transform(df[['loan_amount']])
```

```python
df[['loan_amount','loan_amount_scaled','loan_amount_Mscale','loan_amount_Rscale']].head()
```

| **Technique** | **Main Idea** | **Range** | **Outlier Sensitivity** | **Typical Use** |
|---|---|---|---|---|
| **StandardScaler** | Mean = 0, SD = 1 | No fixed range | High | General-purpose scaling |
| **MinMaxScaler** | Uses minimum and maximum | Usually 0 to 1 | High | Distance-based algorithms |
| **RobustScaler** | Uses median and IQR | No fixed range | **Low** | Data with outliers |
