## 1. Business Problem

Suppose we want to predict an employee's **salary based on years of experience**.

- **Feature (X):** Years of Experience
- **Target (y):** Salary

Our goal is to find a straight line that represents the relationship between experience and salary.


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [30, 35, 42, 45, 50]
})

df
````

````python
plt.scatter(df["Experience"], df["Salary"])
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (₹ in thousands)")
plt.title("Experience vs Salary")
plt.show()
````

## 2. Linear Regression Equation

A simple linear regression model can be represented as:

$$
\hat{y} = b_0 + b_1x
$$

Where:
- $\hat{y}$ = predicted value
- $b_0$ = intercept / constant
- $b_1$ = coefficient / slope
- $x$ = input feature

The model's job is to find the best values of $b_0$ and $b_1$.


## 3. Calculate the Mean of X and Y

First calculate:

$$\bar{x} = \text{mean of X}$$

$$\bar{y} = \text{mean of Y}$$


````python
x = df["Experience"].values
y = df["Salary"].values

x_mean = x.mean()
y_mean = y.mean()

print("Mean of X:", x_mean)
print("Mean of Y:", y_mean)
````

## 4. Calculate the Coefficient / Slope Manually

$$
b_1 = \frac{\sum (x_i-\bar{x})(y_i-\bar{y})}{\sum (x_i-\bar{x})^2}
$$

Create the calculation table to inspect every step.


````python
calculation = pd.DataFrame({
    "X": x,
    "Y": y
})

calculation["X - X_mean"] = calculation["X"] - x_mean
calculation["Y - Y_mean"] = calculation["Y"] - y_mean
calculation["(X-X_mean)(Y-Y_mean)"] = (
    calculation["X - X_mean"] * calculation["Y - Y_mean"]
)
calculation["(X-X_mean)^2"] = calculation["X - X_mean"] ** 2

calculation
````

````python
numerator = calculation["(X-X_mean)(Y-Y_mean)"].sum()
denominator = calculation["(X-X_mean)^2"].sum()

b1 = numerator / denominator

print("Numerator:", numerator)
print("Denominator:", denominator)
print("Coefficient (b1):", b1)
````

## 5. Calculate the Intercept / Constant Manually

$$
b_0 = \bar{y} - b_1\bar{x}
$$


````python
b0 = y_mean - (b1 * x_mean)

print("Intercept (b0):", b0)
````

## 6. Our Regression Equation

Now we have:
- Intercept = $b_0$
- Coefficient = $b_1$

Therefore:

$$
\hat{y} = b_0 + b_1x
$$


````python
print(f"Salary = {b0:.2f} + ({b1:.2f} × Experience)")
````

## 7. Manual Prediction

Suppose a new employee has **6 years of experience**.

Use the regression equation to calculate the predicted salary.


````python
new_experience = 6

manual_prediction = b0 + b1 * new_experience

print(f"Experience: {new_experience} years")
print(f"Predicted Salary: {manual_prediction:.2f} thousand")
````

## 8. Residual / Prediction Error

For each existing observation:

$$
\text{Residual} = \text{Actual} - \text{Predicted}
$$

A good regression line tries to keep the overall prediction error small.


````python
df["Predicted_Salary"] = b0 + b1 * df["Experience"]
df["Residual"] = df["Salary"] - df["Predicted_Salary"]

df
````

## 9. Least Squares

Linear Regression finds the line that minimizes the **sum of squared residuals**.

$$
SSE = \sum (y_i - \hat{y}_i)^2
$$

Squaring the residuals:
- prevents positive and negative errors from cancelling each other
- gives larger errors a greater penalty

The line with the minimum SSE is called the **least-squares regression line**.


````python
sse = (df["Residual"] ** 2).sum()

print("Sum of Squared Errors (SSE):", sse)
````

## 10. Visualize the Regression Line


````python
plt.scatter(df["Experience"], df["Salary"], label="Actual")
plt.plot(
    df["Experience"],
    df["Predicted_Salary"],
    label="Regression Line"
)

plt.xlabel("Experience (Years)")
plt.ylabel("Salary (₹ in thousands)")
plt.title("Linear Regression")
plt.legend()
plt.show()
````

## 11. Implement Linear Regression Using Scikit-learn

Important methods and attributes:
- `fit()` → learns the model parameters
- `coef_` → coefficient/slope
- `intercept_` → intercept/constant
- `predict()` → predicts new values


````python
from sklearn.linear_model import LinearRegression

X = df[["Experience"]]
y = df["Salary"]

model = LinearRegression()
model.fit(X, y)
````

````python
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
````

## 12. Compare Manual Calculation with Scikit-learn

The values learned by Scikit-learn should match our manual calculation.


````python
print("Manual coefficient:", b1)
print("Sklearn coefficient:", model.coef_[0])
print()
print("Manual intercept:", b0)
print("Sklearn intercept:", model.intercept_)
````

## 13. Prediction Using Scikit-learn

Predict the salary for an employee with 6 years of experience.


````python
new_employee = pd.DataFrame({
    "Experience": [6]
})

prediction = model.predict(new_employee)

print(f"Predicted Salary: {prediction[0]:.2f} thousand")
````

## 14. Understanding `fit()` and `predict()`

### `fit(X, y)`

The model learns the relationship between the feature and target.

Conceptually:

```text
X + y
  ↓
Find best coefficient and intercept
  ↓
Store the learned model
```

### `predict(X_new)`

The learned equation is used to calculate predictions for new input values.


## 15. Simple Linear Regression vs Multiple Linear Regression

### Simple Linear Regression

Uses one feature:

$$
\hat{y} = b_0 + b_1x_1
$$

Example:

`Salary = b0 + b1 × Experience`

### Multiple Linear Regression

Uses multiple features:

$$
\hat{y} = b_0 + b_1x_1 + b_2x_2 + b_3x_3 + \cdots
$$

Example:

`Salary = b0 + b1 × Experience + b2 × Age + b3 × Projects`
