| **Aspect**                    | **OrdinalEncoder**                                    | **LabelEncoder**                                                                  |
| ----------------------------- | ----------------------------------------------------- | --------------------------------------------------------------------------------- |
| **Purpose**                   | Encodes **categorical features/columns** into numbers | Encodes **target labels/classes** into numbers                                    |
| **Main use**                  | For **input features (X)**                            | Mainly for **target variable (y)**                                                |
| **Works on**                  | One or multiple feature columns                       | Typically one target column                                                       |
| **Output**                    | 2D array                                              | 1D array                                                                          |
| **Ordering**                  | Can explicitly define meaningful category order       | Assigns numbers based on sorted class labels; order is not necessarily meaningful |
| **Example**                   | `Low → 0, Medium → 1, High → 2`                       | `No → 0, Yes → 1`                                                                 |
| **Best suited for**           | **Ordinal categorical features**                      | **Categorical target variable**                                                   |
| **Can specify custom order?** | ✅ Yes, using `categories=`                            | ❌ No custom ordinal meaning                                                       |
| **Multiple columns?**         | ✅ Yes                                                 | ❌ Designed for a single target array                                              |
| **Typical ML usage**          | `X` preprocessing                                     | `y` preprocessing                                                                 |
| **Scikit-learn module**       | `sklearn.preprocessing.OrdinalEncoder`                | `sklearn.preprocessing.LabelEncoder`                                              |

# Final Encoding

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv("Loan_Dataset.csv")
cat_column = df.select_dtypes(include='object').columns.tolist()

print("*"*50,)
print("Total Categorical Column: ",len(cat_column),"\n")
print("*"*50,)
print("*"*50,"\n","\n")
for i in cat_column:
    print(i, ": ",df[i].nunique())
    print(df[i].value_counts(),"\n")
    print("*"*50,"\n")
```

## Binary Class - Direct Mapping

```python
df["loan_limit"] = df["loan_limit"].map({
    "cf": 0,
    "ncf": 1
})

df["approv_in_adv"] = df["approv_in_adv"].map({
    "nopre": 0,
    "pre": 1
})

df["Credit_Worthiness"] = df["Credit_Worthiness"].map({
    "l1": 0,
    "l2": 1
})

df["open_credit"] = df["open_credit"].map({
    "nopc": 0,
    "opc": 1
})

df["business_or_commercial"] = df["business_or_commercial"].map({
    "nob/c": 0,
    "b/c": 1
})

df["Neg_ammortization"] = df["Neg_ammortization"].map({
    "not_neg": 0,
    "neg_amm": 1
})

df["interest_only"] = df["interest_only"].map({
    "not_int": 0,
    "int_only": 1
})

df["lump_sum_payment"] = df["lump_sum_payment"].map({
    "not_lpsm": 0,
    "lpsm": 1
})

df["construction_type"] = df["construction_type"].map({
    "sb": 0,
    "mh": 1
})

df["Secured_by"] = df["Secured_by"].map({
    "home": 0,
    "land": 1
})

df["co-applicant_credit_type"] = df["co-applicant_credit_type"].map({
    "CIB": 0,
    "EXP": 1
})

df["submission_of_application"] = df["submission_of_application"].map({
    "to_inst": 0,
    "not_inst": 1
})

df["Security_Type"] = df["Security_Type"].map({
    "direct": 0,
    "Indriect": 1
})
```

## Ordinal Encoding

```python
cat_column = df.select_dtypes(include='object').columns.tolist()
print("*"*50,)
print("Total Categorical Column: ",len(cat_column),"\n")
print("*"*50,)
print("*"*50,"\n","\n")
for i in cat_column:
    print(i, ": ",df[i].nunique())
    print(df[i].value_counts(),"\n")
    print("*"*50,"\n")
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[
        ["<25", "25-34", "35-44", "45-54", "55-64", "65-74", ">74"],
        ["1U", "2U", "3U", "4U"]
    ]
)

df[["age", "total_units"]] = encoder.fit_transform(
    df[["age", "total_units"]]
)
```

## One Hot Encoding

```python
cat_column = df.select_dtypes(include='object').columns.tolist()
print("*"*50,)
print("Total Categorical Column: ",len(cat_column),"\n")
print("*"*50,)
print("*"*50,"\n","\n")
for i in cat_column:
    print(i, ": ",df[i].nunique())
    print(df[i].value_counts(),"\n")
    print("*"*50,"\n")
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    drop="first",
    sparse_output=False,
    handle_unknown="ignore"
)

encoded = encoder.fit_transform(df[cat_column])
encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(cat_column),
    index=df.index
)
df = pd.concat(
    [df.drop(columns=cat_column),encoded_df],
    axis=1
)
```
