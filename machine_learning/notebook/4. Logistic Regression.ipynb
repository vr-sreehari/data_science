```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

## Data Loading

```python
df = pd.read_csv('heart_disease.csv')
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

## Data Cleaning

```python
df['education'].value_counts()
```

```python
df['education'] = df['education'].fillna(df['education'].mode()[0])
```

```python
df['education'].isnull().sum()
```

```python
df['cigsPerDay'].value_counts()
```

```python
df['cigsPerDay'] = df['cigsPerDay'].fillna(df['cigsPerDay'].median())
df['cigsPerDay'].isnull().sum()
```

```python
df['totChol'] = df['totChol'].fillna(df['totChol'].median())
df['BMI'] = df['BMI'].fillna(df['BMI'].median())
df['heartRate'] = df['heartRate'].fillna(df['heartRate'].median())
df['glucose'] = df['glucose'].fillna(df['glucose'].median())
df.isnull().sum()
```

```python
df['BPMeds'].value_counts()
```

```python
df['BPMeds'] = df['BPMeds'].fillna(0.0)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.to_csv('heart_disease_cleaned.csv',index=False)
```

## Train Test split

```python
X = df.drop('TenYearCHD', axis=1)
y = df['TenYearCHD']
```

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.20,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
df[df['TenYearCHD']==1]
```

## Scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Model Training

```python
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
Log_model = LogisticRegression()
Log_model.fit(X_train_scaled,y_train)
```

```python
y_predict = Log_model.predict(X_test_scaled)
```

```python
y_predict_proba = Log_model.predict_proba(X_test_scaled)
y_predict_proba
```

```python
for i, j in zip(y_predict.tolist()[51:101],y_predict_proba.tolist()[51:101]):
    print(i,": ",j)
```

```python
accuracy_score(y_test,y_predict)
```
