#Missing Value Imputation

##KNN Impute

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age':[25,28,26,45,47,46],
    'Experience':[2,4,3,20,22,21],
    'Salary':[50000,60000,np.nan,120000,125000,118000]
})

df
```

```python
from sklearn.impute import KNNImputer
```

```python
imputer = KNNImputer(n_neighbors=2)
```

```python
transformed_value = imputer.fit_transform(df)
transformed_value
```

```python
imputed_df = pd.DataFrame(transformed_value,columns=['Age','Experience','Salary'])
imputed_df
```

## Simple Imputation

```python
df4 = df3.copy()
df4
```

### Mean

```python
avg = df4['Age'].mean()
```

```python
df4['Age'] = df4['Age'].fillna(avg)
```

```python
df4.isnull().sum()
```

### Median

```python
df5 = df3.copy()
```

```python
df5['Age'].median()
```

```python
new_median = df5['Age'].median()
df5['Age'] = df5['Age'].fillna(new_median)
```

```python
df5['Age'].isnull().sum()
```

```python
df3[df3['Age'].isnull()]
```

```python
df5[df3['Age'].isnull()]
```

```python
df4[df3['Age'].isnull()]
```

### Mode

```python
import pandas as pd
import numpy as np
df3 = pd.read_csv("/content/Titanic-Dataset.csv")
```

```python
df6 = df3.copy()
```

```python
df6['Cabin'].isnull().sum()
```

```python
df6['Cabin'].value_counts()
```

```python
df6['Cabin'].mode()
```

```python
df6['Cabin'] = df6['Cabin'].fillna(df6['Cabin'].mode()[1])
```

```python
df3[df3['Cabin'].isnull()]
```

```python
df6[df3['Cabin'].isnull()]
```

```python
df6[df3['Cabin'].isnull()]['Cabin'].value_counts()
```

## Group-Wise simple Imputation

```python
df7 = df3.copy()
```

```python
df7.groupby('Sex')['Age'].mean()
```

```python
df7['Age'] = df7['Age'].fillna(df7.groupby('Sex')['Age'].transform('mean'))
```

```python
df7['Age'].isnull().sum()
```

```python
df7[df3['Age'].isnull()]
```
