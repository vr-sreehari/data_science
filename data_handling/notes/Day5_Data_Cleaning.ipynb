## Regex - Regular Expression

```python
df['Phone'].dtype
```

```python
df['New_phone'] = df['Phone'].str.extract(r'(\d{10})')
df[['Phone','New_phone']]
```

### Simple Regex Symbols

- `\d` → a digit
- `+` → one or more
- `{10}` → exactly 10 occurrences
- `()` → capture the matching part
- `$`→ End of the string


```python
data = {'Ord':["ORD-10","ORD-11","ORD-101","ORD-1001"],
        'Name':["Selva","Vinoth","Arun","Shiva"]}
df3 = pd.DataFrame(data)
df3
```
